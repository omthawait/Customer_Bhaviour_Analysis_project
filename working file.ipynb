```python
# fill the missing value of riview coloum 

df["Review Rating"] = df.groupby("Category")["Review Rating"].transform(lambda x : x.fillna(x.median()))

# under this we can impute the missing value with the category group wise by median . using the transform and the lambda function .
```

```python
df.isnull().sum() # now the missing value is fill .
```

```python
df.columns = df.columns.str.lower() # make the all column name into a lower case.
df.columns = df.columns.str.replace(' ','_') # replace (customer id) to (customer_id) all columns.
df = df.rename(columns = {'purchase_amount_(usd)' : 'purchase_amount'}) # rename the pruchase amount coloum.
```

```python
# create a (age_group) coloum

labels = ['Young Adult', 'Adult' , 'Middle-Age' , 'senoir']
df['age_group'] = pd.qcut(df['age'], q=4 , labels = labels)

# using the pd.qcut  ,  they are split he age into 4 group according the data distrubution , lables
# TO UNDER STAND THE CUSTOMER BEHAVIOUR BY AGE_GROUP
```

```python
df[['age' , 'age_group']].head(10)
```

```python
# CREATE COLOUMN PURCHASE FREQUENCY DAYS 
frequency_mapping = {
    'Fortnightly' : 14,
    'Weekly' : 7,
    'Monthly' : 30,
    'Quarterly' : 90,
    'Bi-Weekly': 14,
    'Annually' :365,
    'Every 3 Months' :90 
}

df["purchase_frequency_days"] = df["frequency_of_purchases"].map(frequency_mapping)
```

```python
df[["purchase_frequency_days" , "frequency_of_purchases"]].head(10)
# the bettter under the customer frequency analysis in the formate oof the number and as compare to text , super use for the analysis
```

```python
# remove the promocode coloum 
(df["discount_applied"]== df["promo_code_used"]).all()

# then drop the promodcodeuser coloum

df = df.drop("promo_code_used", axis=1)
```
